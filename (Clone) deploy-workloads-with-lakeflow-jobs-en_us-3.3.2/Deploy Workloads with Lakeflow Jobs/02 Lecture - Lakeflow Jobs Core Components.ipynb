# Lecture - Lakeflow Jobs Core Components

## Overview

In this lecture, you will understand the fundamental building blocks that make up every Lakeflow job and explore how tasks are orchestrated using DAG concepts.

## Learning Objectives

By the end of this lecture, you will be able to:
1. **Understand** the building blocks of Lakeflow Jobs, including the concepts of jobs and tasks
2. **Identify** and **describe** the various task types and configuration options available in Lakeflow Jobs
3. **Explain** how Directed Acyclic Graphs (DAGs) enable task orchestration and workflow patterns
4. **Learn** how to create and configure jobs using the Lakeflow Jobs UI

(function () {
  const root = document.currentScript.previousElementSibling.previousElementSibling;
  const wrap = root && root.classList.contains('jt-wrap') ? root : document.querySelector('.jt-wrap');
  if (!wrap) return;
  const config = {
    jobs: {
      panelId: "panel-jobs",
      highlights: [
        { cls: ".jt-highlight-main", x: 0.0, y: 4.2, w: 100, h: 35 }
      ]
    },
    tasks: {
      panelId: "panel-tasks",
      highlights: [
        { cls: ".task-1", x: 37, y: 22, w: 8, h: 6 },
        { cls: ".task-2", x: 37, y: 37, w: 8, h: 6 },
        { cls: ".task-3", x: 63, y: 30, w: 8, h: 6 }
      ]
    }
  };
  const tabs = wrap.querySelectorAll(".jt-tab");
  const panels = wrap.querySelectorAll(".jt-panel");
  const highlights = wrap.querySelectorAll(".jt-highlight");
  function resetAll() {
    tabs.forEach(tab => {
      tab.classList.remove("active");
      tab.setAttribute("aria-expanded", "false");
      const icon = tab.querySelector(".jt-tab-icon");
      if (icon) icon.textContent = "+";
    });
    panels.forEach(panel => panel.classList.remove("active"));
    highlights.forEach(h => {
      h.style.display = "none";
      h.style.left = "0%";
      h.style.top = "0%";
      h.style.width = "0%";
      h.style.height = "0%";
    });
  }
  function activate(key) {
    resetAll();
    const tab = wrap.querySelector('.jt-tab[data-key="' + key + '"]');
    const panel = wrap.querySelector("#" + config[key].panelId);
    if (!tab || !panel) return;
    tab.classList.add("active");
    tab.setAttribute("aria-expanded", "true");
    tab.querySelector(".jt-tab-icon").textContent = "−";
    panel.classList.add("active");
    config[key].highlights.forEach(item => {
      const el = wrap.querySelector(item.cls);
      if (!el) return;
      el.style.display = "flex";
      el.style.left = item.x + "%";
      el.style.top = item.y + "%";
      el.style.width = item.w + "%";
      el.style.height = item.h + "%";
    });
  }
  tabs.forEach(tab => {
    tab.addEventListener("click", function () {
      const key = this.dataset.key;
      const isActive = this.classList.contains("active");
      if (isActive) {
        resetAll();
      } else {
        activate(key);
      }
    });
  });
})();
</script>

##### EXPAND FOR ADDITIONAL NOTES
<details>

Here are the two most fundamental concepts you need to understand:

A Job is the primary resource for scheduling, coordinating, and running operations such as data processing, ETL, analytics, and machine learning workloads within the Databricks environment. Think of a job as the container that holds your entire workflow.

A task is a single unit of work within a job that executes a specific workload such as a notebook, script, query, and more. Tasks are the individual building blocks that do the actual work.

The relationship is hierarchical: each job consists of one or more tasks, which are the individual units of work that make up the job. The visual shows this clearly - one job containing multiple tasks.

</details>

There are three common workflow patterns you'll encounter:
- Sequence Pattern is used for data transformation, processing, cleaning, and building bronze/silver/gold tables in a medallion architecture.
- Funnel Pattern brings together multiple data sources for data collection and consolidation.
- Fan-out or Star Pattern takes a single data source and distributes it for data ingestion and distribution to multiple downstream systems.

Understanding these patterns helps you design effective workflows for your specific use cases.
